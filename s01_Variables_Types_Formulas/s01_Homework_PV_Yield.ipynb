## Background: radiation, not "sunshine hours"
A PV module converts **global radiation** (direct + diffuse sunlight on a surface, in kWh/m²) into electricity.
Module power is rated at standard test conditions with an irradiance of $G_\text{STC} = 1\ \text{kW/m}^2$.
The annual energy of a system with peak power $P_\text{peak}$ is therefore
$$E = P_\text{peak} \cdot \frac{H}{G_\text{STC}} \cdot PR$$
- $H$: annual global radiation in kWh/m²,
- $H / G_\text{STC}$: the **full-load hours** (or *peak sun hours*): how many hours the module would have to run at
  its rated power to collect $H$,
- $PR$: performance ratio (losses: heat, dust, inverter, cables, orientation), typically 0.75–0.85.

⚠️ Weather services also publish **sunshine duration** (hours per day with strong direct sunlight). That is *not*
the same: Düsseldorf had about 1,700 hours of sunshine in 2023, but only about 1,130 full-load hours of radiation.
