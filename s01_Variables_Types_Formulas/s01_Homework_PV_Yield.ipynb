## Background: how much electricity does a PV system produce?
A **PV module** (photovoltaic module, "solar panel") converts sunlight directly into electricity. The more sunlight
falls on it, the more electricity it produces. Four terms are needed:

**1. Irradiance $G$ — a power per area, in kW/m².** How strongly the sun shines on one square metre *at this
moment*: on a clear summer noon roughly 1 kW/m², under clouds much less, at night zero.

**2. Global radiation $H$ — an energy per area, in kWh/(m²·a).** The sunlight *added up* over a whole year: all the
energy that landed on one square metre, from weak morning light, cloudy days and bright noon sun ("global" = direct
sunlight + diffuse light from the sky; a = annum, year).
Power and energy relate like speed and distance: energy = power × time, e.g. 1 kW for 2 h = 2 kWh.

**3. Rated power — in W or kW ("peak", Wp/kWp).** Every module is tested under the same agreed *Standard Test
Conditions* (STC) with an irradiance of $G_\text{STC} = 1\ \text{kW/m}^2$. The power it delivers in this test is its
rated power, e.g. 430 W. $G_\text{STC}$ is a *reference value*, not the maximum: in reality the irradiance is
mostly lower and sometimes even higher. The rated powers of all modules add up to the peak power $P_\text{peak}$
of the system (in kWp).

**4. Full-load hours $H / G_\text{STC}$ — in h/a.** In reality the irradiance changes all the time. The trick:
how many hours would a *constant* irradiance of $G_\text{STC} = 1\ \text{kW/m}^2$ need to deliver the same energy
$H$ as the real year? Energy ÷ power = time. Example with $H = 1000\ \text{kWh/(m}^2\cdot\text{a)}$:
$$\frac{H}{G_\text{STC}} = \frac{1000\ \text{kWh}/(\text{m}^2\cdot\text{a})}{1\ \text{kW}/\text{m}^2} = 1000\ \text{h/a}
\qquad (\text{m}^2 \text{ cancels},\ \text{kWh} / \text{kW} = \text{h})$$
Like a car that drives 15,000 km in a year at changing speeds: at a constant 100 km/h this would take 150 h.
The power of a module is roughly proportional to the irradiance, so in each of these hours the system delivers
exactly its rated power — hence *full-load hours*. Divided by 365 d/a they give the *peak sun hours* per day.

**Annual energy.** Peak power × full-load hours, reduced by the unavoidable losses:
$$E = P_\text{peak} \cdot \frac{H}{G_\text{STC}} \cdot PR \qquad \text{units: } \text{kW} \cdot \frac{\text{h}}{\text{a}} \cdot 1 = \frac{\text{kWh}}{\text{a}}$$
$PR$ is the *performance ratio* (no unit), typically 0.75–0.85. It covers losses: hot modules produce less, dust,
the inverter (it converts the module's direct current into the alternating current of the house), cables, and a
roof that does not face the sun perfectly.

⚠️ Weather services also publish **sunshine duration** (the hours with strong direct sunlight, per day, month or
year). That is *not* the same: in this region a year has roughly 1,500–1,800 hours of sunshine, but only about
1,130 full-load hours of radiation.
