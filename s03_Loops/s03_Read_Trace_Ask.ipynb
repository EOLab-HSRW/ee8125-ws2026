# Part 0 — Talk about code: line numbers

Turn on line numbers: in JupyterLab menu **View → Show Line Numbers** (or click into a cell and press **Shift + L**).
Now everybody can say "line 3" and we all mean the same line.

How to **read a line aloud**: say what happens, from right to left at the `=` sign.

| Code | Read aloud |
|---|---|
| `s = ""` | "`s` gets an empty string." |
| `for i in range(4):` | "Repeat 4 times; `i` is 0, then 1, 2, 3." |
| `    s = s + "x"` | "Take `s`, add an `x` at the end, store it in `s` again." |
| `    print(s)` | "Show `s` on the screen." |

The lines that start with **4 spaces** (more to the right) belong to the loop. They are the **loop body**: Python
repeats them in every round.

````python
s = ""
for i in range(4):
    s = s + "x"
    print(s)
print("done")
````

# Part A — Glossary: the words we use

Many problems are **word** problems, not code problems. The last column is for **you**: write the word in your own
language, or a short note (double-click this cell to edit it). Add new words whenever you meet one.

| Term | Plain English | Tiny example | My language / my note |
|---|---|---|---|
| value | a piece of data | `30.5`, `"DUI"` | |
| variable | a name for a value | `t = 30.5` | |
| assignment `=` | "store the value on the right under the name on the left" (not "equals"!) | `n = n + 1` | |
| type | the kind of a value | `type(30.5)` → `float` | |
| `int` | whole number, no point | `7`, `-3`, `2023` | |
| `float` | number with a decimal point | `30.5`, `7.0` | |
| `str` (string) | text, in quotes | `"Duisburg"`, `""` (empty) | |
| `bool` | `True` or `False` | `30.5 > 25` → `True` | |
| list | many values in order, in `[ ]` | `[30.5, 28.1, 25.4]` | |

| dict (dictionary) | pairs *key → value*, in `{ }` | `{"DUI": 30.5}` | |
| key | the name you look up in a dict | `d["DUI"]` | |

| iteration / round | one run through the loop body | | |

# Part B — Trace tables: run the code on paper

A **trace table** has one column per variable and one row per round of the loop. You fill it in **before** you run
the code. It is the best way to understand a loop — and in the exam you have no time to guess.

### 🔍 B1 — A worked example
The loop from Part 0:

| round | `i` | `s` before line 3 | `s` after line 3 | printed |
|---|---|---|---|---|
| 1 | 0 | `""` | `"x"` | `x` |
| 2 | 1 | `"x"` | `"xx"` | `xx` |
| 3 | 2 | `"xx"` | `"xxx"` | `xxx` |
| 4 | 3 | `"xxx"` | `"xxxx"` | `xxxx` |

Note: `i` is never used in the body. It only counts the rounds.

You can let Python print the trace for you. This is the **check**, not the first step:

````python
s = ""
for i in range(4):
    before = s
    s = s + "x"
    print("round", i + 1, "  i =", i, "  before:", repr(before), "  after:", repr(s))
````
