## Background: radiation, not "sunshine hours"
A PV module converts **global radiation** (direct + diffuse sunlight on a surface; energy per area over a period, e.g. kWh/m² per year) into electricity.
Module power is rated at standard test conditions with an irradiance of $G_\text{STC} = 1\ \text{kW/m}^2$.
The annual energy of a system with peak power $P_\text{peak}$ is therefore
$$E = P_\text{peak} \cdot \frac{H}{G_\text{STC}} \cdot PR$$
- $H$: annual global radiation in kWh/(m²·a) (a = annum, year),
- $H / G_\text{STC}$: the **full-load hours** in h/a: how many hours per year the module would have to run at
  its rated power to collect $H$ (units: kWh/(m²·a) ÷ kW/m² = h/a). Divided by 365 d/a this gives the
  *peak sun hours* per day,
- $PR$: performance ratio (losses: heat, dust, inverter, cables, orientation), typically 0.75–0.85.

⚠️ Weather services also publish **sunshine duration** (the hours with strong direct sunlight, per day, month or
year). That is *not* the same: in this region a year has roughly 1,500–1,800 hours of sunshine, but only about
1,130 full-load hours of radiation.
