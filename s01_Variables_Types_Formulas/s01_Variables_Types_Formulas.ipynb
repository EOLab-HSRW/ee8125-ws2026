# Part 1 — Python as a calculator

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `+` `-` | add, subtract | `7 - 2` | `5` |
| `*` | multiply | `3 * 4` | `12` |
| `/` | divide (always gives a decimal number) | `7 / 2` | `3.5` |
| `//` | integer division (rounds down) | `7 // 2` | `3` |
| `%` | remainder (modulo) | `7 % 2` | `1` |
| `**` | power | `2 ** 10` | `1024` |

Precedence as in maths: `**` before `*` `/` before `+` `-`. Use parentheses when in doubt.

````python
print(7 / 2)
print(7 // 2)
print(7 % 2)
print(2 ** 10)
````

````python
# Precedence: what is the mean of 12 and 18?
print(12 + 18 / 2)      # wrong: division first
print((12 + 18) / 2)    # right
````

````python
# 135 minutes are how many hours and minutes?
print(135 // 60, "h", 135 % 60, "min")
````

# Part 2 — Variables

A **variable** is a name for a value. `=` means *assign* (store the value on the right under the name on the left),
not *equal*. The right side is evaluated first.

````python
area_m2 = 250 * 80        # store
area_ha = area_m2 / 10_000
print(area_ha, "ha")
````

````python
print(area_m2)
````

````python
DU = 500000
KL = 38500
KL == DU
````

````python
# Reassignment: the right side is evaluated first, then stored
rain_total_mm = 0
rain_total_mm = rain_total_mm + 4.2
rain_total_mm = rain_total_mm + 1.8
print(rain_total_mm)
````

````python
# Short form: += -= *= /=
rain_total_mm += 3.0
print(rain_total_mm)
````

**Rules and good habits for names**
- letters, digits, `_`; must not start with a digit; case matters (`T` ≠ `t`);
- no Python keywords (`if`, `for`, `class`, `lambda`, …);
- use descriptive names with units: `temp_max_c`, `wind_speed_ms`, `area_m2` — not `x`, `a1`, `t2`.

# Part 4 — Formulas from physics and environment

Typical work in this course: a formula from a lecture or a data sheet → Python → number with unit.

**Example: power in the wind.** The power of the wind flowing through the rotor area $A$ of a turbine is
$$P_{\rm Wind} = \cdot \tfrac{1}{2}\,\rho\,A\,v^3$$
with air density $\rho \approx 1.225$ kg/m³ and wind speed $v$ in m/s. A real turbine converts at most
59.3 % (Betz limit), in practice about 45 %.

````python
import math   # a "toolbox" with pi, sqrt, sin, log, ... (more in session 7)

rho_kgm3 = 1.225
rotor_diameter_m = 120
wind_speed_ms = 8.
efficiency = 0.45

area_m2 = math.pi * (rotor_diameter_m / 2) ** 2
p_wind_w = 0.5 * rho_kgm3 * area_m2 * wind_speed_ms ** 3
p_el_w = efficiency * p_wind_w
print(p_el_w / 1e6, "MW")
````

````python
import matplotlib.pyplot as plt
fig, ax = plt.subplots()
ax.plot(wind_speed_ms,p_el_w,".")
````

````python
import numpy as np
x = np.arange(0,5.01,0.1)
x = np.array([0.,1.,4,5,10])
y = x**2
print(x)
print(y)
````

````python
# v³ matters: double the wind speed -> 8 times the power
wind_speed_ms = 16.0
print(0.5 * rho_kgm3 * area_m2 * wind_speed_ms ** 3 * efficiency / 1e6, "MW")
````

# Part 5 — Printing results: f-strings

An **f-string** (`f"…"`) puts values into text: `{name}` is replaced by the value.
After a colon you can format: `{x:.1f}` one decimal, `{x:,.0f}` thousands separator, `{x:8.2f}` width 8, `{x:.2e}` scientific, `{x:.0%}` percent.

````python
1e-3
````

````python
print(f"The el. power is {p_el_w:014.2f} W or {p_el_w:014.2e} W")
````

````python
station = "Kleve"
temp_c = 21.4567
p_el_w = 2_294_536.7
share = 0.4512
print(f"{station}: {temp_c:.1f} °C")
print(f"Power: {p_el_w:,.0f} W = {p_el_w / 1e6:.1f} MW = {p_el_w:.1e} W")
print(f"Share: {share:.0%}")
````

````python
# Aligned columns: < left, > right, number = width
print(f"{'Station':<15}{'T / °C':>8}")
print(f"{'Kamp-Lintfort':<15}{21.4:>8.1f}")
print(f"{'Kleve':<15}{19.75:>8.1f}")
````

# Part 7 — Reading error messages · ⭐ Extra

Errors are normal. Read the **last line** first: it names the error type and the reason.
The arrow `---->` shows the line where it happened. The cells below fail on purpose.

````python
temperature = "Kamp-Lintfort"
````

````python
print(temperature)        # NameError: this name was never defined (typo? cell not run?)
````

````python
 28 + "Elevation: "       # TypeError: str + int is not defined
````

````python
days = 0
rain_per_day = 25 / days  # ZeroDivisionError
````

````python
float("21,4")             # ValueError: Python uses the decimal POINT
````

| Error | Typical reason |
|---|---|
| `SyntaxError` | typo in the code: missing `)` or `"`, `=` instead of `==`, … |
| `NameError` | name not defined: typo, or the defining cell was not run |
| `TypeError` | operation does not fit the types, e.g. `str + int` |
| `ValueError` | right type, wrong content, e.g. `float("abc")` |
| `ZeroDivisionError` | division by zero |
