# Part 1 — Booleans and comparisons
| Operator | Meaning |
|---|---|
| `==` / `!=` | equal / not equal (`=` assigns, `==` compares!) |
| `<` `<=` `>` `>=` | less / greater (or equal) |
| `in` / `not in` | contained in a list, string, dict keys, set |

The result is `True` or `False` (type `bool`).

````python
tmin_c = -2.4
print(tmin_c < 0)
print(tmin_c == -2.4)        # (Duisburg-Baerl, night of 7 February 2023)
print("KLE" in ["KLF", "KLE"], "rain" in "light rain")
print(0 <= 0.45 <= 1)           # chained: 0 <= x and x <= 1
````

````python
# Never compare floats with == after calculating
x = 0.1 + 0.2
print(x == 0.3, abs(x - 0.3) < 1e-9)
````

# Part 2 — `if`, `elif`, `else`
```python
if condition:
    runs only if condition is True
elif other_condition:          # optional, as many as you like
    runs if the first was False and this one is True
else:                          # optional
    runs if all conditions were False
```
Exactly **one** branch runs: the first whose condition is `True`.

````python
tmin_c = -2.4
if tmin_c < 0:
    print("Frost warning: protect your plants!")
print("checked")
````

````python
precip_mm = 3.4
if precip_mm > 0:
    print("wet day")
else:
    print("dry day")
````

**Classification.** The EU air quality index classes PM2.5 (daily mean, µg/m³) roughly like this:
≤ 10 good, ≤ 20 fair, ≤ 25 moderate, ≤ 50 poor, above very poor. Order matters: test from low to high.

````python
pm25 = 26.0      # Düsseldorf-Lörick, 21 January 2023 (Umweltbundesamt)
if pm25 <= 10:
    category = "good"
elif pm25 <= 20:
    category = "fair"
elif pm25 <= 25:
    category = "moderate"
elif pm25 <= 50:
    category = "poor"
else:
    category = "very poor"
print(f"PM2.5 {pm25} µg/m³: {category}")
````

| Expression | True if |
|---|---|
| `a and b` | both are true |

| `not a` | `a` is false |

# Part 4 — Loops + `if`: count, filter, classify
Most data processing is a loop with an `if` inside.

````python
tmax = [31.6, 32.2, 31.8, 31.4, 29.9, 31.7, 30.1, 25.2, 20.5, 21.4]    # Duisburg-Baerl, 5–14 September 2023
# count
hot_days = 0
for t in tmax:
    if t >= 30:
        hot_days += 1
print("hot days (≥ 30 °C):", hot_days)

# filter: keep the matching values
hot = []
for t in tmax:
    if t >= 30:
        hot.append(t)
print(hot)
# the same as a comprehension with a condition
print([t for t in tmax if t >= 30])
print(sum(1 for t in tmax if t >= 25), "summer days (≥ 25 °C)")
````

````python
# find the maximum and its position by hand (what max() does inside)
best = tmax[0]
best_i = 0
for i, t in enumerate(tmax):
    if t > best:
        best = t
        best_i = i
print(f"max {best} °C on day {best_i + 1}")
````

````python
# classify and count in a dictionary
counts = {"cool": 0, "summer": 0, "hot": 0}
for t in tmax:
    if t >= 30:
        counts["hot"] += 1
    elif t >= 25:
        counts["summer"] += 1
    else:
        counts["cool"] += 1
print(counts)
````

### 🔍 Step by step: which branch runs?
The classifying loop above does three things for every value: **test**, run **one** branch, **update** the dictionary.
We take it apart: *What is it? – One first – Trace – Short form.* Predict before every cell.

````python
# What is it? – ONE value, both tests
t = tmax[0]                   # 31.6
print(t, t >= 30, t >= 25)    # both tests are True ...
if t >= 30:
    print("hot")              # ... but only the first true branch runs
elif t >= 25:
    print("summer")
````

````python
# Trace: the loop from above, with a print in every round
counts = {"cool": 0, "summer": 0, "hot": 0}
for t in tmax:
    if t >= 30:
        branch = "hot"
    elif t >= 25:
        branch = "summer"
    else:
        branch = "cool"
    counts[branch] += 1
    print(f"t = {t:4.1f}  t>=30: {t >= 30!s:5}  t>=25: {t >= 25!s:5}  -> {branch:6}  {counts}")
````

For 31.6 both tests are `True`, but the `elif` is never checked. What if the wider test comes first? Predict, then run:

````python
counts_wrong = {"cool": 0, "summer": 0, "hot": 0}
for t in tmax:
    if t >= 25:               # wrong order: the wider test first
        counts_wrong["summer"] += 1
    elif t >= 30:             # never reached for a hot day – it is already "summer"
        counts_wrong["hot"] += 1
    else:
        counts_wrong["cool"] += 1
print(counts_wrong)
````

# Part 5 — `break` and `continue`
- `break` leaves the loop immediately (e.g. "find the first …").
- `continue` skips the rest of this round and goes to the next item (e.g. "ignore invalid values").

Loggers often write a code such as `-999` for missing values.

````python
readings = [12.1, 12.4, -999, 12.8, 13.5, -999, 14.2]     # example values; DWD files use -999, too
valid = []
for r in readings:
    if r == -999:
        continue              # skip the missing value
    valid.append(r)
print(valid, "mean:", round(sum(valid) / len(valid), 2))
````

````python
# first day on which the cumulative rain reaches 30 mm
rain_mm = [0.0, 3.4, 12.1, 0.0, 0.2, 5.8, 0.0, 22.5, 1.0]
total = 0.0
for day, r in enumerate(rain_mm, start=1):
    total += r
    if total >= 30:
        print("30 mm reached on day", day)
        break
else:                          # runs only if the loop did NOT break
    print("30 mm not reached")
````

# Part 6 — Checking input
Users type anything. A `while True` loop asks **until** the answer is valid, then `break`s.
(Typed text that is not a number makes `float()` crash — session 6 shows how to catch that with `try`/`except`.)

````python
while True:
    answer = input("Save the report? (y/n) ").strip().lower()
    if answer in ("y", "n"):
        break
    print("Please type y or n.")
print("saved" if answer == "y" else "not saved")
````

### 🔍 Step by step: how does `while True` end?
`while True` never stops by itself — only `break` ends it. To *see* the rounds, we replace `input()` by a list of
answers a user might type, and take one answer per round with `pop(0)` (session 2).

````python
# What is it? – the test and the cleaning of one answer
print("maybe" in ("y", "n"), "y" in ("y", "n"))
print(repr(" Y ".strip().lower()))
````

````python
# Trace: three answers, one per round
answers = ["maybe", "", " Y "]
attempt = 0
while True:
    attempt += 1
    answer = answers.pop(0).strip().lower()      # instead of asking the user
    print(f"round {attempt}: answer = {answer!r}, valid: {answer in ('y', 'n')}")
    if answer in ("y", "n"):
        print("   -> break: leave the loop")
        break
    print("   -> not valid: back to 'while True'")
print("after the loop, answer =", repr(answer))
````

# Part 7 — Plot it: colours by category
Loop + `if` can also choose a **colour** for every bar. `ax.bar(x, y, color=colors)` accepts a list with one colour
per bar. (Same recipe as in session 3: `fig, ax = plt.subplots(...)`, draw, label.)

````python
import matplotlib.pyplot as plt

tmax = [21.0, 22.6, 23.8, 28.1, 31.6, 32.2, 31.8, 31.4, 29.9, 31.7, 30.1, 25.2, 20.5, 21.4, 24.3,
        26.1, 26.2, 24.7, 20.3, 23.6, 22.7, 19.0, 17.9, 21.0, 22.6, 23.9, 25.6, 22.2, 23.7, 19.9]
# daily maxima, Duisburg-Baerl, 1–30 September 2023 (DWD)
days = list(range(1, len(tmax) + 1))
colors = []
for t in tmax:
    if t >= 30:
        colors.append("tab:red")          # hot day
    elif t >= 25:
        colors.append("tab:orange")       # summer day
    else:
        colors.append("tab:gray")

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=150)
ax.bar(days, tmax, color=colors)
ax.axhline(30, color="tab:red", ls="--", lw=1)
ax.axhline(25, color="tab:orange", ls="--", lw=1)
ax.set(xlabel="day of September 2023", ylabel="daily maximum temperature in °C",
       title="Duisburg-Baerl: a late heatwave (red ≥ 30 °C, orange ≥ 25 °C; data: DWD)")
plt.show()
````

# Summary
- Conditions: `== != < <= > >=`, `in`, combined with `and`, `or`, `not`.
- `if` / `elif` / `else`: the **first** true branch runs — order the tests carefully.
- Loop + `if` = count, filter, classify, find extremes. Comprehension with filter: `[x for x in xs if cond]`.
- `break` = stop the loop, `continue` = next item.
- Validate input in a `while True` loop.
