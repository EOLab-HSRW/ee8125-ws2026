# Part 1 — `for` loops
```python
for item in collection:
    do something with item      # indented block (4 spaces) = loop body
after the loop                  # not indented: runs once, after the loop
```
The loop variable (`item`) takes each value in turn. Choose a name that says what it is: `for temp in temps`.

````python
temps_c = [23.8, 22.9, 20.9, 21.4, 21.2]      # daily means, Duisburg-Baerl, 12–16 June 2023
for temp in temps_c:
    print(temp, "°C =", round(temp + 273.15, 2), "K")
print("done")
````

### 1.1 The accumulator pattern
1. create the result **before** the loop, 2. update it **inside**, 3. use it **after** the loop.

````python
rain_mm = [0.0, 0.0, 8.7, 0.4, 22.0, 0.2, 0.5]      # Duisburg-Baerl, 16–22 October 2023
total = 0.0                       # 1. start value
for r in rain_mm:
    total = total + r             # 2. update
print(f"total: {total:.1f} mm")   # 3. use
````

````python
# Building a new list
temps_f = []
for temp in temps_c:
    temps_f.append(temp * 9 / 5 + 32)
print(temps_f)
````

````python
# Cumulative rainfall: a running total stored at every step
cumulative = []
running = 0.0
for r in rain_mm:
    running += r
    cumulative.append(round(running, 1))
print(cumulative)
````

**Indentation matters.** Compare: what is inside the loop runs 7 times, what is outside once.

````python
count = 0
for r in rain_mm:
    count += 1
    print("inside, count =", count)
print("outside, count =", count)
````

# Part 2 — `range()`: counting and simulating
`range(stop)`, `range(start, stop)`, `range(start, stop, step)` — like slicing, `stop` is not included.

````python
print(list(range(5)))
print(list(range(2, 8)))
print(list(range(2000, 2031, 10)))
````

**Simulating year by year.** The atmospheric CO₂ concentration was about 419 ppm in 2023 (annual mean at Mauna Loa,
NOAA) and grew by about 2.5 ppm per year over the last decade. A loop over the years gives a simple projection.

````python
co2_ppm = 419.3
growth_ppm = 2.5
for year in range(2024, 2031):
    co2_ppm += growth_ppm
    print(year, round(co2_ppm, 1))
````

Growth by a *percentage* compounds: each year starts from the new value.

````python
pv_capacity_gw = 82.0     # installed PV in Germany, end of 2023 (Bundesnetzagentur, approx.)
rate = 0.15               # +15 % per year
for year in range(2024, 2031):
    pv_capacity_gw *= 1 + rate
    print(f"{year}: {pv_capacity_gw:6.1f} GW")
````

# Part 3 — `zip()` and `enumerate()`
`zip(a, b)` walks through two (or more) lists **in parallel**. `enumerate(a)` gives you the position **and** the item.

````python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
tmax = [30.5, 28.1, 25.4, 27.8, 27.6, 28.5, 28.7]
tmin = [16.0, 16.1, 15.8, 12.7, 11.8, 10.6, 13.9]    # Duisburg-Baerl, 12–18 June 2023
for day, hi, lo in zip(days, tmax, tmin):
    print(f"{day}: {lo:5.1f} … {hi:5.1f} °C   range {hi - lo:4.1f} K")
````

````python
for i, day in enumerate(days, start=1):
    print(i, day)
````

````python
# Day-to-day change: compare each value with the next one
changes = []
for today, tomorrow in zip(tmax[:-1], tmax[1:]):
    changes.append(round(tomorrow - today, 1))
print(changes)
````

# Part 4 — Loops over dictionaries, nested loops
A `for` loop over a dictionary gives the **keys**; `.items()` gives key and value together.

````python
elevation_m = {"Kleve": 46, "Geldern-Walbeck": 37, "Duisburg-Baerl": 24, "Düsseldorf": 37, "Essen-Bredeney": 150}
for name in elevation_m:
    print(name)
for name, elev in elevation_m.items():
    print(f"{name:<15}{elev:>4} m")
````

A loop inside a loop runs the inner loop completely for **each** step of the outer loop.

````python
weekly_tmax = {"DUI": [30.5, 28.1, 25.4], "ESS": [28.4, 26.4, 23.7], "KLE": [30.8, 28.6, 26.1]}   # 12–14 June 2023
for sid, values in weekly_tmax.items():
    line = sid + ":"
    for v in values:
        line += f" {v:5.1f}"
    print(line, f"  mean {sum(values) / len(values):.1f}")
````

# Part 5 — `while` loops
`while condition:` repeats as long as the condition is `True`. Use it when you **don't know in advance** how many
steps are needed. Conditions use comparisons — more about them next week:

`<` `<=` `>` `>=` `==` (equal) `!=` (not equal) → result is `True` or `False`.

````python
print(3 < 5, 419.3 >= 450, "KLE" == "DUS", 2 != 3)
````

````python
# When does CO2 reach 450 ppm at +2.5 ppm per year?
co2_ppm = 419.3
year = 2023
while co2_ppm < 450:
    year += 1
    co2_ppm += 2.5
print(year, round(co2_ppm, 1))
````

````python
# Radioactive decay: Cs-137 has a half-life of 30.1 years. After how many half-lives is less than 1 % left?
fraction = 1.0
n = 0
while fraction >= 0.01:
    fraction /= 2
    n += 1
print(n, "half-lives =", round(n * 30.1), "years, fraction left:", fraction)
````

⚠️ If the condition never becomes `False`, the loop runs forever. Stop it with the ■ button (*Kernel → Interrupt*).
Always make sure something in the loop body moves towards the end.

# Part 7 — Looping over the lines of a file · ⭐ Extra
A text file can be read line by line with a `for` loop. `data/temps_week.txt` holds the daily maxima of
Duisburg-Baerl, 12–18 June 2023 (DWD, CC BY 4.0), one value per line. `with open(...)` closes the file automatically afterwards.
Each line is a **string** that ends with a newline `\n` — `strip()` removes it, `float()` converts.
(Files in detail: session 6.)

````python
with open("data/temps_week.txt", encoding="utf-8") as f:
    for line in f:
        print(repr(line))          # repr shows the hidden \n
````

````python
values = []
with open("data/temps_week.txt", encoding="utf-8") as f:
    for line in f:
        values.append(float(line.strip()))
print(values)
print(f"mean {sum(values) / len(values):.1f} °C, max {max(values)} °C")
````

# Part 8 — Plot it: a first figure
Numbers in a list become much clearer in a figure. **matplotlib** draws figures; today we use it as a *recipe* —
copy the pattern, change the data and the labels. What `fig` and `ax` are exactly, you learn in session 7.
```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=150)    # an empty figure (8 × 4.5 inches) with one plot area
ax.plot(x, y)                                         # a line through the points; ax.bar(x, y) for bars
ax.set(xlabel="...", ylabel="... in unit", title="...")
plt.show()
```
`x` and `y` are lists of the same length. Every axis label gets a **unit**, every figure a **source**.

````python
import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
rain_mm = [0.0, 0.0, 8.7, 0.4, 22.0, 0.2, 0.5]      # Duisburg-Baerl, 16–22 October 2023 (DWD)

cumulative = []                                     # the accumulator pattern from Part 1
running = 0.0
for r in rain_mm:
    running += r
    cumulative.append(running)

fig, ax = plt.subplots(figsize=(8, 4.5), dpi=150)
ax.bar(days, rain_mm, label="daily")
ax.plot(days, cumulative, marker="o", color="tab:red", label="cumulative")
ax.set(ylabel="precipitation in mm", title="Duisburg-Baerl, 16–22 October 2023 (data: DWD)")
ax.legend()
plt.show()
````

# Summary
| Pattern | Code |
|---|---|
| every item | `for x in xs:` |
| n times / a sequence of numbers | `for i in range(start, stop, step):` |
| two lists in parallel | `for a, b in zip(xs, ys):` |
| position and item | `for i, x in enumerate(xs, start=1):` |
| keys and values | `for k, v in d.items():` |
| until a condition is false | `while condition:` |
| new list from old | `[f(x) for x in xs]` |
| a first figure | `fig, ax = plt.subplots(figsize=(8, 4.5), dpi=150)`, `ax.plot(x, y)`, `ax.bar(x, y)`, `ax.set(...)` |
