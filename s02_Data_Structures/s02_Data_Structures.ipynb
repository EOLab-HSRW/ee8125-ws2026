# Part 1 — Lists
A **list** is an ordered collection in square brackets. It can grow and shrink, and its items can change.

````python
tmax_c = [30.8, 28.6, 26.1, 28.0, 28.8, 29.4, 26.3]     # daily maxima, Kleve, Mon 12 – Sun 18 June 2023
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
mixed = ["Kleve", 46, True, 6.0954]                      # possible, but rarely a good idea
empty = []
print(tmax_c)
print(len(tmax_c), len(empty))
````

### 1.1 Indexing — counting starts at 0
```
days     "Mon"  "Tue"  "Wed"  "Thu"  "Fri"  "Sat"  "Sun"
index      0      1      2      3      4      5      6
negative  -7     -6     -5     -4     -3     -2     -1
```

````python
print(days[0], tmax_c[0])
print(days[2], tmax_c[2])
print(days[-1], tmax_c[-1])     # last item
````

````python
days[7]                          # IndexError: there is no 8th item
````

### 1.2 Slicing — `list[start:stop:step]`
`stop` is **not** included. Missing `start` means "from the beginning", missing `stop` "to the end".

````python
print(tmax_c[0:5])      # Mon..Fri (index 0,1,2,3,4)
print(tmax_c[:5])       # same
print(tmax_c[5:])       # weekend
print(tmax_c[-3:])      # last three
print(tmax_c[::2])      # every second day
print(tmax_c[::-1])     # reversed
````

### 1.3 Changing lists

````python
readings = [12.1, 12.4, 99.9, 12.8]
readings[2] = 12.6             # replace an item (99.9 was a sensor error)
readings.append(13.0)          # add at the end
readings.insert(0, 11.8)       # add at position 0
print(readings)
last = readings.pop()          # remove and return the last item
readings.remove(12.4)          # remove the first item with this value
print(readings, "removed:", last)
````

````python
# Combining lists
week1 = [3.2, 0.0, 1.5]
week2 = [0.0, 7.8]
print(week1 + week2)           # new list
week1.extend(week2)            # week1 itself is changed
print(week1)
````

### 1.4 Useful functions for lists of numbers

````python
print("max:", max(tmax_c), " min:", min(tmax_c), " sum:", round(sum(tmax_c), 1))
print("mean:", round(sum(tmax_c) / len(tmax_c), 2))
print("sorted:", sorted(tmax_c))                 # new sorted list
print("descending:", sorted(tmax_c, reverse=True))
print("hottest day:", days[tmax_c.index(max(tmax_c))])   # .index() = position of a value
print("28.0 measured?", 28.0 in tmax_c)
print("count of 0.0 in rain list:", [0.0, 3.1, 0.0, 0.0, 2.2].count(0.0))
````

`sorted(x)` returns a new list; `x.sort()` sorts `x` itself and returns `None`.

# Part 2 — Tuples
A **tuple** is like a list in round brackets, but it **cannot be changed** (immutable).
Use it for values that belong together and stay together: coordinates, a date (year, month, day), a record.

````python
kleve = (51.7612, 6.0954)          # (lat, lon) of the DWD station Kleve
print(kleve[0], len(kleve))
````

````python
kleve[0] = 52.0                    # TypeError: tuples cannot be changed
````

````python
# Unpacking: one name per item
lat, lon = kleve
print(f"lat = {lat}°N, lon = {lon}°E")
# Swap two values without a helper variable
a, b = 1, 2
a, b = b, a
print(a, b)
````

# Part 3 — Dictionaries
A **dictionary** maps **keys** to **values**: `{key: value, ...}`. You look up by key, not by position —
like a real dictionary or a phone book. Keys are unique (usually strings or numbers).

````python
station = {"id": "KLE", "dwd_id": "02629", "name": "Kleve", "lat": 51.7612, "lon": 6.0954, "elevation_m": 46}
print(station["name"], station["elevation_m"])
print(len(station))
````

````python
station["height"]                  # KeyError: no such key
````

````python
print(station.get("height"))           # None instead of an error
print(station.get("height", "n/a"))    # with a default
print("elevation_m" in station)        # checks the KEYS
````

````python
# Changing: assign to a key (new key = add, existing key = overwrite)
station["setting"] = "rural"
station["elevation_m"] = 45      # pretend a new survey
del station["lon"]
print(station)
````

````python
print(list(station.keys()))
print(list(station.values()))
print(list(station.items()))       # (key, value) tuples
````

# Part 4 — Sets
A **set** holds each value **only once** and has no order. Good for: unique values, membership tests, comparing groups.

````python
observed = ["rain", "sun", "rain", "fog", "sun", "rain"]
print(set(observed))
print(len(set(observed)), "different weather types")
````

````python
# Not every weather station measures everything (DWD, 2023):
sunshine_stations = {"GEL", "DUS", "ESS"}     # Geldern-Walbeck, Düsseldorf, Essen-Bredeney
wind_stations = {"DUS", "ESS"}
all_stations = {"KLE", "GEL", "DUI", "DUS", "ESS"}
print("sunshine and wind:", sunshine_stations & wind_stations)    # intersection
print("sunshine or wind: ", sunshine_stations | wind_stations)    # union
print("only sunshine:    ", sunshine_stations - wind_stations)    # difference
print("neither:          ", all_stations - (sunshine_stations | wind_stations))
print("KLE has wind?", "KLE" in wind_stations)
````

# Part 5 — Strings are sequences too
Indexing, slicing, `len` and `in` work on strings as well. Strings are immutable, like tuples.
`split()` cuts a string into a **list**, `join()` glues a list of strings together.

````python
date = "2023-06-14"
print(date[0:4], date[5:7], date[-2:])      # year, month, day
print(len(date), "-06-" in date)

line = "2023-10-20,KLE,13.5,22.1"
parts = line.split(",")
print(parts)
tmax = float(parts[2])
print(tmax + 1)

print(" | ".join(["KLE", "DUI", "DUS"]))
````

# Part 6 — Names and objects: a common trap
`b = a` does **not** copy a list. It gives the *same* list a second name. Changing it through one name
changes what you see through the other. Use `.copy()` (or `list(a)`, `a[:]`) for an independent copy.

````python
a = [1, 2, 3]
b = a            # same list, two names
c = a.copy()     # a new list
b.append(4)
print("a:", a, " b:", b, " c:", c)
print(a is b, a is c)
````

Numbers, strings and tuples are immutable, so this cannot surprise you with them.

### Which container when?
| Need | Use |
|---|---|
| ordered values, may change (a time series) | `list` |
| fixed group of values (coordinates, a record) | `tuple` |
| look up by a name or ID (metadata, factors) | `dict` |
| unique values, compare groups | `set` |

# Summary
| | list | tuple | dict | set |
|---|---|---|---|---|
| literal | `[1, 2]` | `(1, 2)` | `{"a": 1}` | `{1, 2}` |
| ordered | yes | yes | insertion order | no |
| changeable | yes | no | yes | yes |
| access | `x[0]`, `x[1:3]` | `x[0]` | `x["a"]`, `.get()` | `in` |
