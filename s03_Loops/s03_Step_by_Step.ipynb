# Part A — A dictionary of lists, a loop in a loop

```python
weekly_tmax = {"DUI": [30.5, 28.1, 25.4], "ESS": [28.4, 26.4, 23.7], "KLE": [30.8, 28.6, 26.1]}   # 12–14 June 2023
```

### 🔍 A1 — What is it?
Predict first: what does each line print? What type is each thing?

```python
print(type(weekly_tmax))
print(weekly_tmax["DUI"])          # the content of ONE compartment ...
print(type(weekly_tmax["DUI"]))    # ... is an ordinary list
print(weekly_tmax["DUI"][0])       # the first value in this list
```

`weekly_tmax["DUI"]` is just a list — everything from Part 1 works with it. Give it a name:

```python
values = weekly_tmax["DUI"]
print(values, len(values), sum(values) / len(values))
```

### 🔍 A2 — Only the outer loop
`.items()` gives key **and** value. Predict: how many lines are printed? What are `sid` and `values` in each round?

```python
for sid, values in weekly_tmax.items():
    print("sid =", sid, "   values =", values)
```

### 🔍 A3 — Only the inner loop, for ONE station
No outer loop yet: we pick one compartment by hand.

```python
values = weekly_tmax["DUI"]
for v in values:
    print("  v =", v)
```

### 🔍 A4 — Both together, with a trace
Now the outer loop goes **around** A3. Predict: how many `inner:` lines? In which order?

```python
for sid, values in weekly_tmax.items():
    print("outer:", sid)
    for v in values:
        print("    inner:", v)
    print("outer done with", sid)
print("all done")
```

The inner loop runs **completely** for every step of the outer loop: 3 stations × 3 values = 9 inner lines.
Indentation tells Python what belongs to which loop.

### 🔍 A5 — Back to the short form (the cell from `s03_Loops`, Part 4)
`line` is an accumulator (Part 1.1): it starts **inside the outer loop** (new for every station) and grows **inside the
inner loop**. The extra `print` shows it growing.

```python
for sid, values in weekly_tmax.items():
    line = sid + ":"                    # 1. start value – for every station anew
    for v in values:
        line += f" {v:5.1f}"            # 2. update – for every value
        print("    line so far:", repr(line))
    print(line, f"  mean {sum(values) / len(values):.1f}")   # 3. use – once per station
```

# Part B — `enumerate(zip(...))`
Homework 3, Task 1 asks for a **numbered** table of **two** lists. `zip` pairs the lists, `enumerate` numbers the pairs.

```python
days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]
tmax = [30.5, 28.1, 25.4, 27.8, 27.6, 28.5, 28.7]
tmin = [16.0, 16.1, 15.8, 12.7, 11.8, 10.6, 13.9]    # Duisburg-Baerl, 12–18 June 2023
```

### 🔍 B1 — What is it?
`list(...)` makes the pairs visible. Predict first: what are the elements of each list?

```python
print(list(zip(days, tmax)))              # pairs (tuples): ('Mon', 30.5), ...
print(list(enumerate(days, start=1)))     # pairs: (1, 'Mon'), ...
print(list(enumerate(zip(days, tmax), start=1)))   # (1, ('Mon', 30.5)): a number and a pair
```

Every element of the last list is a pair whose **second part is again a pair**: `(1, ('Mon', 30.5))`.

### 🔍 B2 — One first: unpack one element by hand

```python
first = list(enumerate(zip(days, tmax), start=1))[0]
print(first)
i, pair = first          # outer pair: number and inner pair
day, t = pair            # inner pair: day and temperature
print(i, day, t)
i, (day, t) = first      # both in one line: the brackets copy the shape of (1, ('Mon', 30.5))
print(i, day, t)
```

### 🔍 B3 — Trace, then back to the short form
The loop header does exactly what B2 did by hand — in every round.

```python
for item in enumerate(zip(days, tmax), start=1):
    print("item =", item)

for i, (day, t) in enumerate(zip(days, tmax), start=1):
    print(f"{i}  {day}  {t:5.1f} °C")
```

# Summary
| Pattern | Step by step |
|---|---|
| dictionary of lists | `d["DUI"]` is a list → name it (`values = d["DUI"]`) → solve for one → loop with `.items()` around it |
| loop in a loop | the inner loop runs completely for every outer step; indentation decides what belongs where |
| `enumerate(zip(a, b))` | elements look like `(1, (a0, b0))` → unpack with `for i, (x, y) in ...` |

**The habit:** *What is it? – One first – Trace – Short form.* Use it whenever code looks too dense to read.
